### Problem 3

Set `p3` to the value of $t$ for which the following matrix is singular:
$$\begin{bmatrix} 1 & 2 & 0 \\ 0 & 3 & 1 \\ 0 & 6 & t \end{bmatrix}.$$

```python
# YOUR CHANGES HERE

# define a function to get det 
def det_ar(t):
    M=np.array([[1,2,0],[0,3,1],[0,6,t]],dtype=float)
    return np.linalg.det(M)

# t appears only once, so det changes in a straight line as t changes
d0= det_ar(0)
d1=det_ar(1)
k=-d0/(d1-d0)   
p3 = round(k)
```

```python
p3
```
